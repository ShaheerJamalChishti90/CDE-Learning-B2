```python
# if age < 10 then 25%
# if age 20 & 50 then 10 % 
# if age > 50 then 50%
```

```python
def apply_discount(age):
    if age < 10 :
        return 0.25
    elif 20 <= age <= 50:
        return 0.1
    elif age > 50:
        return 0.5
            
titanic['Discount'] = titanic['Age'].apply(apply_discount)
titanic['Discount Fare'] = titanic['Fare'] * (1-titanic['Discount']) # 1 - 0.1 == 0.9
titanic[['Name','Fare','Age','Discount','Discount Fare']]
```

```python
def pct_30(column):
    return column.quantile(0.3)
def pct_40(column):
    return column.quantile(0.4)
def pct_75(column):
    return column.quantile(0.75)


def pct_value(column, value):
    return column.quantile(value)

titanic['Age'].agg(pct_value,value=0.4)
# titanic['Age'].agg([pct_30,pct_40,pct_75])
# titanic['Age'].agg(lambda x: x.quantile(0))
```

```python
titanic[['Age', 'Fare']].agg(lambda x: x.quantile(0.3))
```

Pivot tables


```python
#pivot and implicitly define agffunc=np.mean
titanic.pivot_table(values = 'Age', index='Sex')
```

```python
#explicitly define statistics i:e np.median
titanic.pivot_table(values= 'Age', index='Sex', aggfunc=np.median)
```

```python
#multiple statistics
titanic.pivot_table(values='Age', index='Sex', aggfunc=[np.std, np.median,np.mean])
```

```python
#pivot on two varibales
titanic.pivot_table(values='Age', index='Sex', columns='Survived')
# aggfunc = np.mean
```

```python
titanic.pivot_table(values='Age', index='Sex', columns='Survived', fill_value=0)
```
